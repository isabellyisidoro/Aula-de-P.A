```python
# Nome em maiúsculo Crie: um programa que solicite ao usuário seu nome e mostre esse nome completamente em letras maiúsculas.

nome = "Theo"
print(nome.upper())
```

```python
 #Nome em minúsculo: Crie um programa que solicite o nome de uma cidade e exiba o nome todo em letras minúsculas.

cidade = "Mauá"
print(cidade.lower())
```

```python
#Primeira letra maiúscula :Peça ao usuário para digitar seu nome completo e exiba o nome utilizando a primeira letra maiúscula.

nome = input("Digite seu nome completo: ")
print(nome.capitalize())
```

```python
#Organizando nomes Peça ao usuário para digitar uma frase com todas as letras minúsculas. Depois, mostre a frase com a primeira letra de cada palavra maiúscula.

frase = input("Digite uma frase com todas as letras minúsculas:")
print(frase.title())
```

```python
#Substituindo palavras: Peça ao usuário para digitar a frase: Eu gosto de estudar Java. Utilize o comando adequado para substituir a palavra Java por Python e mostre o resultado.


frase = input("Digite: Eu gosto de estudar Java. ")
frase = frase.replace("Java", "Python")
print(frase)
linguagem = "Eu estudo Java"
print(linguagem.replace("Java", "Python"))
```

```python
#Crie um programa que solicite ao usuário digitar seu nome com espaços antes e depois. Utilize o comando adequado para remover esses espaços e exiba o resultado.

nome = input("Digite seu nome com espaços: ")

print(nome.strip())
```

```python
 #Crie um programa que solicite: n Nome do aluno; n Nome da escola; n Nome do curso. Depois, mostre: n Nome do aluno em letras maiúsculas; n Nome da escola em letras minúsculas; n Nome do curso com a primeira letra de cada palavra maiúscula.

nome = input("Digite o nome do aluno: ")
escola = input("Digite o nome da escola: ")
curso = input("Digite o nome do curso: ")

print(nome.upper())
print(escola.lower())
print(curso.title())
```
