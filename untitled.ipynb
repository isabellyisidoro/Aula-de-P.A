#Exercício 4

```python
#Peça um número ao usuário e faça uma contagem até 0, mostrando cada número na tela.

numero = int(input("Digite um número: "))
contador = 1
soma = 0

while contador <= numero:
    soma += contador
    contador += 1

print(f"a Soma é {soma}")
```

#Exercício 8

```python
#peça números vários números ao usuário até que ele digite "0".

numero = int(input("Digite um número: "))
maior = numero

while numero != 0:
    if numero > maior:
        maior = numero
    numero = int(input("Digite um número: "))

if maior != 0:
    print(f"O Maior número é {maior}")
else:
    print("Nenhum número válido foi digitado.")
```

#exercício 9

```python
#faça um programa que peça 5 notas de um aluno usando while.

contador = 1
soma = 0.0

while contador <= 5:
    nota = float(input(f"Digite a nota {contador}: "))
    soma += nota
    contador += 1

media = soma / 5
print(f" A média é {media:.1f}")
```

#Exercício 10

```python
#faça um programa que comece com um saldo de 1000.00.

saldo = 1000.00
opcao = 0

while opcao != 4:
    print("\n--- MENU CAIXA ELETRÔNICO ---")
    print("1 - Depositar")
    print("2 - Sacar")
    print("3 - Ver saldo")
    print("4 - Sair")
    opcao = int(input("Escolha uma opção: "))

    if opcao == 1:
        valor = float(input("Valor do depósito: R$ "))
        saldo += valor
        print(f"Depósito de R$ {valor:.2f} realizado com sucesso!")
    elif opcao == 2:
        valor = float(input("Valor do saque: R$ "))
        if valor <= saldo:
            saldo -= valor
            print(f"Saque de R$ {valor:.2f} realizado com sucesso!")
        else:
            print("Saldo insuficiente!")
    elif opcao == 3:
        print(f"Saldo atual: R$ {saldo:.2f}")
    elif opcao == 4:
        print("Saindo... Obrigado por utilizar o caixa eletrônico!")
    else:
        print("Opção inválida! Tente novamente!")
```
