```python
#Exercício 6 — Guardando uma coluna em uma lista Crie uma lista vazia chamada coluna. Percorra a matriz, guarde nela os números da segunda coluna e, ao final, exiba a lista.

matriz = [[3, 6, 9],[12, 15, 18],[21, 24, 27]]

coluna = []
for linha in matriz:
 coluna.append(linha[2])

print(coluna)
```

```python
#Exercício 8 — Somando uma coluna Crie um programa que some todos os números da segunda coluna e exiba o resultado.

matriz = [

[10, 20, 30],

[40, 50, 60],

[70, 80, 90]

]
soma = 0

for linha in matriz:
    soma += linha[1]

print("a soma da segunda coluna é:", soma)
```

```python
#Exercício 9 — Encontrando o maior número Utilize dois comandos for para percorrer a matriz e encontrar o maior número. Ao final, exiba o resultado.

matriz = [

[12, 5, 8],

[20, 3, 15],

[7, 25, 10]

]
maior = matriz[0][0]

for linha in matriz:
 for numero in linha:
  if numero > maior: maior = numero

print("o maior número é:", maior)
```
