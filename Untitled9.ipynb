```python
# excerício 3: operações matemáticas
#Solicite dois numeros e apresente os seguintes resultados :
#soma,subtração,multiplicaçao e divisão

numero1 = float(input("Digite o primeiro numero:"))
numero2 = float(input("Digite o segundo número:"))

soma = numero1 + numero2
subtração = numero1 - numero2
multiplicação = numero1 * numero2
divisão = numero1 / numero2

print("Soma:", soma)
print("subtração:",subtração)
print("multiplicação:", multiplicação)
print("divisão:", divisão)
```

```python
#Exercício 4: Média de notas
#solicite três notas de um aluno
#calcule a media e a informe a situação do aluno conforme os critérios:
#Média maior ou igual a 7:Aprovado
#Média entre 5 e 6,9: recuperação
#Média menor que 5:Reprovado

nota1 = float(input("Digite a primeira nota: "))
nota2 = float(input("Digite a segunda nota: "))
nota3 = float(input("Digite a terceira nota: "))
# Calcula a média
media = (nota1 + nota2 + nota3) / 3

print(f"Média: {media:5}")

if media >= 7:
    print("Situação: Aprovado")
elif media >= 5:
    print("Situação: Recuperação")
else:
    print("Situação: Reprovado")
```

```python
numero1 = float(input("Digite o primeiro número: "))
numero2 = float(input("Digite o segundo número: "))

operacao = input("Digite a operação (+, -, * ou /): ")

if operacao == "+":
    resultado = numero1 + numero2
elif operacao == "-":
    resultado = numero1 - numero2
elif operacao == "*":
    resultado = numero1 * numero2
elif operacao == "/":
    resultado = numero1 / numero2
else:
    print("Operação inválida.")
    resultado = None

if resultado is not None:
    print("Resultado:", resultado)
```
