```python
#Exercício 1 – Idade para dirigir
#Receba a idade de uma pessoa e diga se ela pode tirar carteira de motorista.
#A pessoa pode tirar a carteira se tiver idade maior ou igual a 18 anos.

idade = int(input("Digite sua idade:"))
if idade >= 18:
  print("Pode tirar carteira de motorista.")
else:
  print("Não pode tirar carteira de motorista.")
```

```python
#Exercício 4 – Situação do aluno
#Receba a nota de um aluno e mostre sua situação:
#Nota maior ou igual a 7 → Aprovado
#Nota maior ou igual a 5 e menor que 7 → Recuperação
#Nota menor que 5 → Reprovado

nota = float(input("Digite a nota do aluno: "))

if nota >= 7:
    print("Aprovado")
elif nota >= 5:
    print("Recuperação")
else:
    print("Reprovado")


```

```python
#Exercício 5 – Desconto na compra
#Receba o valor de uma compra. Se o valor for maior ou igual a 100, aplique 10% de
#desconto e mostre o valor final. Caso contrário, mostre o valor normal da compra.

compra = float(input("Digite o valor total da compra: R$ "))

if compra > 100:
    desconto = compra - (compra * 0.10)
    print (f"o valor a pagar é {desconto}")

elif compra > 500:
    desconto = compra - (compra * 0.20)
    print (f"o valor a pagar é: {desconto}")

else:
   print  (f"você não obteve desconto, valor a pagar : {compra}")
```

```python
#Exercício 7 – Classificação de idade
#Receba a idade de uma pessoa e classifique-a de acordo com sua faixa etária:
# Menor de 12 anos → Criança
# De 12 a 17 anos → Adolescente
# De 18 a 59 anos → Adulto
# 60 anos ou mais → Idoso

idade = int(input("Digite sua idade:"))
if idade < 13:
  print ("Criança")
elif idade <= 18:
  print ("Adolecente")
elif idade <= 59:
  print ("Adulto")
else:
  print ("Idoso")
```

```python
6#Exercício 8 – Média de duas notas
#Receba duas notas, calcule a média e diga se o aluno foi aprovado ou reprovado. O aluno
#será aprovado se a média for maior ou igual a 7.

nota1 = float(input("Digite a primeira nota: "))
nota2 = float(input("Digite a segunda nota: "))
nota3 = float(input("Digite a terceira nota: "))
# Calcula a média
media = (nota1 + nota2 + nota3) / 3

print(f"Média: {media:5}")

if media >= 7:
    print("Situação: Aprovado")
elif media >= 5:
    print("Situação: Recuperação")
else:
    print("Situação: Reprovado")
```

```python
#Exercício 10 – Calculadora de IMC
#Receba o peso e a altura de uma pessoa, calcule o IMC usando a fórmula:
#IMC = peso / (altura × altura)
#Depois, classifique o resultado:
# IMC menor que 18.5 → Abaixo do peso
# IMC de 18.5 a 24.9 → Peso normal
# IMC de 25 a 29.9 → Sobrepeso
# IMC maior ou igual a 30 → Obesidade

# Entrada de dados
peso = float(input("Digite o peso em kg: "))
altura = float(input("Digite a altura em metros: "))

# Cálculo do IMC
imc = peso / (altura * altura)

# Exibição do resultado do IMC formatado com duas casas decimais
print(f"\nSeu IMC é: {imc:.2f}")

# Classificação do IMC com base nas faixas informadas
if imc < 18.5:
    print("Classificação: Abaixo do peso")
elif 18.5 <= imc <= 24.9:
    print("Classificação: Peso normal")
elif 25.0 <= imc <= 29.9:
    print("Classificação: Sobrepeso")
else:
    print("Classificação: Obesidade")
```
